```python
import google.generativeai as genai
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
genai.configure(api_key=os.getenv('GEMINI_API_KEY1'))
```

```python
model=genai.GenerativeModel('gemini-2.5-flash')
```

```python
from IPython.display import Markdown
import textwrap
def demo(text):
  return Markdown(textwrap.indent(text,'>'))
```

```python
while True:
    content=input('enter your content').lower()
    if content=='exit':
        print('good bye')
        break

    prompt  =f'answer to the content clearly within 100 words :{content}'
    response1=model.generate_content(prompt)
    print(response1.text)
```

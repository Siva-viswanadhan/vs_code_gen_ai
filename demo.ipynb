```python
import google.generativeai as genai
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
genai.configure(api_key=os.getenv('GEMINI_API_KEY'))
```

```python
model=genai.GenerativeModel('gemini-2.5-flash')
```

```python
response=model.generate_content("Who is the greatest manager in football history,simple answer")        #for asking queston=generate_content()
```

```python
from IPython.display import Markdown
import textwrap
def demo(text):
  return Markdown(textwrap.indent(text,'>'))
```

```python
demo(response.text)
```

```python
content=input('enter what you want to know in one word')
prompt  =f'answer to the content clearly within 100 words{content}'
response1=model.generate_content(prompt)
demo(response1.text)
```
